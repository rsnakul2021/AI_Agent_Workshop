## 1. Variables & Data Types

A **variable** is a name that stores a value.

Python figures out the type for you (you don't declare `int x` like in some languages).

```python
# Numbers
age = 25                 # int (whole number)
temperature = 36.6       # float (decimal)

# Text
name = "Nakul"           # str

# True / False
is_student = True        # bool

# "Nothing" value
result = None            # NoneType

print(name, age, temperature, is_student, result)
print(type(name), type(age), type(temperature), type(is_student))
```

```python
# Basic arithmetic
a, b = 10, 3
print("add:", a + b)
print("sub:", a - b)
print("mul:", a * b)
print("div:", a / b)      # always float in Python 3
print("floor div:", a // b)
print("modulo:", a % b)   # remainder
print("power:", a ** b)
```

## 2. Strings

Strings are text. In AI work you will pass **lots** of strings (prompts, documents, responses).

```python
model = "gpt-4.1"
prompt = "Summarize this paragraph in 2 bullets."

# Concatenation
print("Using model: " + model)

# f-strings (preferred)
print(f"Using model: {model}")
print(f"Prompt length: {len(prompt)} characters")

# Useful string methods
text = "  LangChain is useful  "
print(text.strip())          # remove surrounding spaces
print(text.lower())
print(text.upper())
print("Chain" in text)       # membership check
print(text.replace("useful", "powerful").strip())
```

```python
# Multi-line strings (great for prompts)
system_prompt = """
You are a helpful AI assistant.
Be concise and accurate.
If you don't know, say you don't know.
""".strip()

print(system_prompt)
```

## 3. Collections: list, tuple, set, dict

| Type | Ordered? | Mutable? | Typical use |
|---|---|---|---|
| `list` | Yes | Yes | sequences of items |
| `tuple` | Yes | No | fixed records |
| `set` | No | Yes | unique items |
| `dict` | Yes (keys) | Yes | key → value maps |

```python
# LIST — ordered, changeable
tools = ["search", "calculator", "code_runner"]
tools.append("web_browser")
print(tools)
print(tools[0])       # first item
print(tools[-1])      # last item
print(tools[0:2])     # slicing
```

```python
# TUPLE — ordered, immutable
point = (10, 20)
print(point[0], point[1])

# SET — unique values only
tags = {"rag", "agent", "rag", "llm"}
print(tags)  # duplicates removed
```

```python
# DICT — key/value pairs (extremely common in AI apps)
user_message = {
    "role": "user",
    "content": "What is an AI agent?"
}

print(user_message["role"])
print(user_message.get("content"))
print(user_message.get("temperature", 0.0))  # default if key missing

user_message["content"] = "Explain RAG in one sentence."
print(user_message)
```

## 5. Loops (`for` and `while`)

```python
models = ["gpt-4.1", "gpt-4o-mini", "llama3.2"]

for m in models:
    print(f"Available model: {m}")

print("---")

for i, m in enumerate(models, start=1):
    print(f"{i}. {m}")
```

```python
# Looping over a dictionary
config = {"temperature": 0.0, "max_tokens": 256, "model": "gpt-4.1"}

for key, value in config.items():
    print(f"{key} = {value}")
```

```python
# while loop — careful with infinite loops!
retries = 3
while retries > 0:
    print(f"Trying... retries left: {retries}")
    retries -= 1
print("Done")
```

## 6. Functions

A **function** packages reusable logic.

In AI apps, you constantly write small functions: format prompts, call tools, clean text, etc.

```python
def greet(name):
    return f"Hello, {name}! Welcome to the AI workshop."

print(greet("Aisha"))
print(greet("Rohan"))
```

```python
# Default arguments + keyword arguments
def build_prompt(question, style="concise", language="English"):
    return (
        f"Answer in {language}. Style: {style}.\n"
        f"Question: {question}"
    )

print(build_prompt("What is LangChain?"))
print("---")
print(build_prompt("What is LangChain?", style="detailed", language="simple English"))
```

```python
# Functions can return multiple values (as a tuple)
def analyze_text(text):
    words = text.split()
    return len(text), len(words)

chars, words = analyze_text("AI agents use tools to complete goals")
print(f"chars={chars}, words={words}")
```

## 7. Intermediate Python (very useful in AI code)

### 7.1 List / dict comprehensions

```python
docs = ["Intro to RAG", "Vector Databases", "Agent Loops", "Prompt Design"]

# List comprehension
titles_upper = [d.upper() for d in docs]
short_titles = [d for d in docs if len(d) < 15]

print(titles_upper)
print(short_titles)

# Dict comprehension
lengths = {d: len(d) for d in docs}
print(lengths)
```

### 7.2 Unpacking

```python
person = ("Ada", "Engineer", 28)
name, role, age = person
print(name, role, age)

first, *middle, last = [1, 2, 3, 4, 5]
print(first, middle, last)
```

### 7.3 `*args` and `**kwargs`

- `*args` → extra positional arguments (tuple)
- `**kwargs` → extra keyword arguments (dict)

You'll see this style in many libraries (including LangChain wrappers).

```python
def log_event(event_name, *args, **kwargs):
    print("Event:", event_name)
    print("args:", args)
    print("kwargs:", kwargs)

log_event("model_call", "gpt-4.1", 0.0, temperature=0, max_tokens=200)
```

### 7.4 Lambda (tiny anonymous functions)

```python
scores = [{"name": "A", "score": 88}, {"name": "B", "score": 95}, {"name": "C", "score": 70}]

# Sort by score descending
ranked = sorted(scores, key=lambda x: x["score"], reverse=True)
print(ranked)
```

## 8. A bit more advanced (still beginner-friendly)

### 8.1 Exceptions (`try` / `except`)

Real AI apps fail sometimes (network, missing keys, bad JSON). Handle errors instead of crashing.

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

```python
# Common AI pattern: validate config
def get_api_key(env: dict):
    key = env.get("OPENAI_API_KEY")
    if not key:
        raise ValueError("OPENAI_API_KEY is missing. Add it to your environment/.env")
    return key

try:
    get_api_key({})
except ValueError as e:
    print("Caught error:", e)
```

### 8.2 Type hints

Type hints don't change runtime behavior much — they make code clearer (and help editors/tools).

```python
def format_chat_message(role: str, content: str) -> dict:
    return {"role": role, "content": content}

msg: dict = format_chat_message("user", "Explain tokens in LLMs")
print(msg)
```

### 8.3 Classes (objects)

A **class** is a blueprint. An **object** is one instance.

You don't need deep OOP for Day 1, but LangChain code is full of classes (`ChatOpenAI`, chains, agents).

```python
class SimpleAgent:
    def __init__(self, name: str, tools: list[str]):
        self.name = name
        self.tools = tools

    def introduce(self) -> str:
        tool_list = ", ".join(self.tools)
        return f"I am {self.name}. I can use: {tool_list}"

    def has_tool(self, tool_name: str) -> bool:
        return tool_name in self.tools


agent = SimpleAgent("ResearchBot", ["search", "summarize"])
print(agent.introduce())
print("Can search?", agent.has_tool("search"))
print("Can code?", agent.has_tool("code_runner"))
```

### 8.4 Decorators (light intro)

A decorator wraps a function to add behavior (logging, timing, retries).

You will *see* `@something` in libraries even if you rarely write your own at first.

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        elapsed = time.time() - start
        print(f"{func.__name__} took {elapsed:.4f}s")
        return result
    return wrapper

@timer
def fake_model_call(prompt: str) -> str:
    time.sleep(0.2)  # pretend network latency
    return f"Echo: {prompt}"

print(fake_model_call("Hello agent"))
```

### 8.5 Working with JSON (critical for APIs)

```python
import json

payload = {
    "model": "gpt-4.1",
    "messages": [
        {"role": "system", "content": "You are helpful."},
        {"role": "user", "content": "What is an agent?"},
    ],
}

# Python dict → JSON string (sending to an API)
json_text = json.dumps(payload, indent=2)
print(json_text)

# JSON string → Python dict (reading API response)
parsed = json.loads(json_text)
print(parsed["messages"][1]["content"])
```
